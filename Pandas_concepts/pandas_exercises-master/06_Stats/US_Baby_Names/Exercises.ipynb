### Step 5. Delete the column 'Unnamed: 0' and 'Id'

```python
baby_names.drop(columns=["Unnamed: 0","Id"],inplace=True)
baby_names
```

### Step 6. What year has the highest number of baby names in the dataset?

```python
# maxBabyCount=baby_names.Count.max()
# baby_names.loc[baby_names["Count"]==maxBabyCount,"Year"]

df=baby_names.groupby("Year").Count.sum()
df=df.to_frame()
print(df.Count.idxmax())
```

### Step 7. Is there more male or female names in the dataset?

```python
baby_names.Gender.value_counts()
```

### Step 8. Group the dataset by name and assign to names

```python
babygroup=baby_names.groupby("Name")["Name"].value_counts()
babygroup
```

### Step 9. How many different names exist in the dataset?

```python
# totalnames=pd.Series(baby_names.Name.unique())
# totalnames=totalnames.to_frame()
# totalnames.shape
print(f"total unique names : {baby_names.Name.unique().shape[0]}")
```

### Step 10. What is the name with most occurrences?

```python
mostCommonName=baby_names.Name.value_counts()
mostCommonName=mostCommonName.sort_values(ascending=False).to_frame()
mostCommonName.iloc[0]
```

### Step 11. How many different names have the least occurrences?

```python
mini=mostCommonName["count"].min()
mostCommonName[mostCommonName["count"]==mini]
```

### Step 12. What is the median name occurrence?

```python
dfmid=baby_names.groupby("Name").Count.median()
dfmid=dfmid.to_frame()
dfmid
```
